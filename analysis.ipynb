```python
import numpy as np
import pandas as pd

def read_xvg(file_path, start_time=500):
    """
    Reads RMSD data from a GROMACS .xvg file and returns the RMSD values after a specified start time.
    
    Parameters:
    - file_path (str): Path to the .xvg file.
    - start_time (float): The time threshold after which data should be collected.
    
    Returns:
    - np.array: Array of RMSD values after the specified start time.
    """
    with open(file_path, 'r') as file:
        lines = file.readlines()

    rmsd_values = []
    for line in lines:
        if line.strip() and not line.startswith(('#', '@')):
            parts = line.split()
            time = float(parts[0])
            if time >= start_time:
                rmsd = float(parts[1])
                rmsd_values.append(rmsd)
    
    return np.array(rmsd_values)

# File paths to the three .xvg files
#file1 = '/Users/chenyifang/Desktop/RA/gaa/md/r1/rmsd_lig.xvg'
file2 = '/Users/chenyifang/Desktop/RA/gaa/md/r2/rmsd_TM_BB.xvg'
file3 = '/Users/chenyifang/Desktop/RA/gaa/md/r3/rmsd_TM_BB.xvg'

# Read RMSD data from the .xvg files after 500ns
#rmsd1 = read_xvg(file1, start_time=500)
rmsd2 = read_xvg(file2, start_time=500)
rmsd3 = read_xvg(file3, start_time=500)

# Combine the RMSD data from all three replicas
all_data = np.concatenate((rmsd2, rmsd3))

# Convert to a DataFrame
df = pd.DataFrame(all_data, columns=['RMSD'])

# Export to CSV
df.to_csv('/Users/chenyifang/Desktop/RA/gaa/md/rmsd_TM_BB_500ns.csv', index=False)
```

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

def read_xvg(file_path):
    """
    Reads RMSD data from a GROMACS .xvg file and returns the time and RMSD values.
    """
    with open(file_path, 'r') as file:
        lines = file.readlines()

    times = []
    rmsd_values = []
    for line in lines:
        if line.strip() and not line.startswith(('#', '@')):
            parts = line.split()
            time = float(parts[0])
            rmsd = float(parts[1])
            times.append(time)
            rmsd_values.append(rmsd)
    
    return np.array(times), np.array(rmsd_values)

# File paths to the three .xvg files for the first protein
file1_protein1 = '/Users/chenyifang/Desktop/RA/SLC6A8/r1/rmsd_TM_BB.xvg'
file2_protein1 = '/Users/chenyifang/Desktop/RA/SLC6A8/r2/rmsd_TM_BB.xvg'
file3_protein1 = '/Users/chenyifang/Desktop/RA/SLC6A8/r3/rmsd_TM_BB.xvg'

# File paths to the three .xvg files for the second protein
file1_protein2 = '/Users/chenyifang/Desktop/RA/rgx/r1/rmsd_TM_BB.xvg'
file2_protein2 = '/Users/chenyifang/Desktop/RA/rgx/r2/rmsd_TM_BB.xvg'
file3_protein2 = '/Users/chenyifang/Desktop/RA/rgx/r3/rmsd_TM_BB.xvg'

# Read RMSD data from the .xvg files for the first protein
times1_protein1, rmsd1_protein1 = read_xvg(file1_protein1)
times2_protein1, rmsd2_protein1 = read_xvg(file2_protein1)
times3_protein1, rmsd3_protein1 = read_xvg(file3_protein1)

# Combine the RMSD data from all three replicas for the first protein
all_times_protein1 = np.concatenate((times1_protein1, times2_protein1, times3_protein1))
all_rmsd_protein1 = np.concatenate((rmsd1_protein1, rmsd2_protein1, rmsd3_protein1))

# Read RMSD data from the .xvg files for the second protein
times1_protein2, rmsd1_protein2 = read_xvg(file1_protein2)
times2_protein2, rmsd2_protein2 = read_xvg(file2_protein2)
times3_protein2, rmsd3_protein2 = read_xvg(file3_protein2)

# Combine the RMSD data from all three replicas for the second protein
all_times_protein2 = np.concatenate((times1_protein2, times2_protein2, times3_protein2))
all_rmsd_protein2 = np.concatenate((rmsd1_protein2, rmsd2_protein2, rmsd3_protein2))

# Number of bins for the histogram
n_bins = 20

# Create the combined plot with two subplots
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(15, 6))

# Plot RMSD over time for the first protein
ax1.plot(times1_protein1, rmsd1_protein1, color='#85C1E9', alpha=0.6, label='Protein 1 - Run 1')
ax1.plot(times2_protein1, rmsd2_protein1, color='#85C1E9', alpha=0.6, label='Protein 1 - Run 2')
ax1.plot(times3_protein1, rmsd3_protein1, color='#85C1E9', alpha=0.6, label='Protein 1 - Run 3')

# Plot RMSD over time for the second protein
ax1.plot(times1_protein2, rmsd1_protein2, color='#F1948A', alpha=0.6, label='Protein 2 - Run 1')
ax1.plot(times2_protein2, rmsd2_protein2, color='#F1948A', alpha=0.6, label='Protein 2 - Run 2')
ax1.plot(times3_protein2, rmsd3_protein2, color='#F1948A', alpha=0.6, label='Protein 2 - Run 3')

# Add labels and title for the first subplot
ax1.set_xlabel('Time (ns)')
ax1.set_ylabel('RMSD (nm)')
ax1.set_title('RMSD Over Time')
ax1.legend()

# Create the probability density plot on the second subplot
sns.histplot(all_rmsd_protein1, bins=n_bins, kde=True, color='#85C1E9', label='Protein 1', ax=ax2)
sns.histplot(all_rmsd_protein2, bins=n_bins, kde=True, color='#F1948A', label='Protein 2', ax=ax2)

# Invert the axes to match the desired orientation
ax2.invert_xaxis()
ax2.set_xlabel('Probability Density')
ax2.set_ylabel('RMSD (nm)')
ax2.set_title('Probability Distribution of RMSD')
ax2.legend()

# Adjust layout for better spacing
plt.tight_layout()

# Show the plot
plt.show()

```

```python
import pandas as pd

# Define the input and output file paths
input_csv_path = '/Users/chenyifang/Desktop/RA/SLC6A8/rmsd_TM_BB_500ns.csv'
output_colvar_path = '/Users/chenyifang/Desktop/RA/SLC6A8/COLVAR'

# Read the CSV file
df = pd.read_csv(input_csv_path)

# Check if the CSV file contains a column named "RMSD"
if 'RMSD' not in df.columns:
    raise ValueError("The input CSV file must contain a column named 'RMSD'")

# Create a new DataFrame for the COLVAR format
colvar_df = pd.DataFrame()

# Add a time column starting from 1 with the same length as the RMSD column
colvar_df['time'] = range(1, len(df) + 1)

# Add the RMSD column
colvar_df['rmsd'] = df['RMSD']

# Save the DataFrame to a COLVAR file
colvar_df.to_csv(output_colvar_path, sep=' ', index=False, header=True)

print(f"Successfully converted {input_csv_path} to {output_colvar_path}")
```
