```python
n=17
column=1

bfactor =[]

with open('rmsf-Phd.xvg','r') as f:
    lines=f.readlines()
    
    for row in lines[n:]:
        columns=row.split()
        if len(columns) >1:
            bfactor.append(eval(columns[column]))
bfactor
for residue, bf in zip(structure.get_residues(), bfactor):
    for atom in residue:
        atom.set_bfactor(bf)

io = PDBIO()
io.set_structure(structure)
io.save("Phd_raw_bfactor.pdb")      
    
```
